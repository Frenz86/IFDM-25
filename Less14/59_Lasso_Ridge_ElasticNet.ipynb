# Regularization in Linear Regression

Regularization is a technique used in linear regression to prevent overfitting by adding a penalty term to the cost function. It discourages the model from fitting the training data too closely and helps in creating simpler models that generalize well to new, unseen data.

<img src="https://frenzy86.s3.eu-west-2.amazonaws.com/python/elastic.png" width=1200 >


## Lasso Regression (L1 Regularization)

Lasso regression adds the absolute values of the coefficients as a penalty term to the cost function. The formula for Lasso regression is:

$J(\theta) = \frac{1}{2m} \sum_{i=1}^{m} (h_\theta(x^{(i)}) - y^{(i)})^2 + \lambda \sum_{j=1}^{n} |\theta_j|$

- $J(\theta) $: Cost function.
- m: Number of training examples.
- $h_\theta(x^{(i)})$: Hypothesis function.
- $y^{(i)}$: Actual output for the ith example.
- $\theta_j$: Coefficients/weights.
- $\lambda$: Regularization parameter.

The key characteristic of Lasso is that it tends to shrink some coefficients to exactly zero, effectively performing feature selection.

## Ridge Regression (L2 Regularization)

Ridge regression adds the squared values of the coefficients as a penalty term to the cost function. The formula for Ridge regression is:

$J(\theta) = \frac{1}{2m} \sum_{i=1}^{m} (h_\theta(x^{(i)}) - y^{(i)})^2 + \lambda \sum_{j=1}^{n} \theta_j^2 $

The term \( \lambda \sum_{j=1}^{n} \theta_j^2 \) penalizes large coefficients, but unlike Lasso, Ridge does not lead to sparsity; all features contribute to the model, although with smaller weights.

## Elastic Net (L1+L2)

Elastic Net is a combination of L1 and L2 regularization. It adds both the absolute values and squared values of the coefficients to the cost function. The formula for Elastic Net is a combination of the Lasso and Ridge terms:

$ J(\theta) = \frac{1}{2m} \sum_{i=1}^{m} (h_\theta(x^{(i)}) - y^{(i)})^2 + \lambda_1 \sum_{j=1}^{n} |\theta_j| + \lambda_2 \sum_{j=1}^{n} \theta_j^2 $

- $ \lambda_1 $: L1 regularization parameter.
- $ \lambda_2 $: L2 regularization parameter.

Elastic Net provides a balance between Lasso and Ridge, combining the feature selection property of Lasso with the robustness to multicollinearity offered by Ridge.


### Quando usare cosa?

- Ridge: quando pensi che molte variabili siano rilevanti
- Lasso: quando vuoi interpretabilità e selezione automatica delle variabili
- Elastic Net: il compromesso tra i due

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings('ignore')
```

```python
#path = "boston_houses.csv"
path = "https://frenzy86.s3.eu-west-2.amazonaws.com/IFAO/boston_houses.csv"

data = pd.read_csv(path)
data.rename(columns={'MEDV': 'Price'}, inplace=True)
data
```

```python
## STEP1 DEFINIZIONE DI INPUT E OUTPUT
target = 'Price'
X = data.drop(columns=[target])
y = data[target]
```

```python
## STEP2 TRAIN TEST SPLIT
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y,
                                                    test_size=0.3,
                                                    random_state=667,
                                                   )
```

```python
## STEP3 MODELLAZIONE (REGRESSION)
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train,y_train)
```

```python
## STEP4 PREDIZIONE
y_pred = model.predict(X_test)
```

```python
## STEP5 VERIFICA METRICHE (for Regression)
from sklearn.metrics import r2_score,mean_absolute_error,root_mean_squared_error,mean_squared_error

r2 = r2_score(y_test,y_pred)
mse = mean_squared_error(y_test,y_pred)
mae = mean_absolute_error(y_test,y_pred)
rmse = root_mean_squared_error(y_test,y_pred)

print(f'R2 score {r2}')
print(f'MSE score {mse}')
print(f'MAE score {mae}')
print(f'RMSE score {rmse}')
```

```python
from sklearn.model_selection import cross_val_score

r2_score = cross_val_score(model, X, y, scoring='r2', cv=5)
r2_score
#np.mean(r2_score)
```

```python
model.coef_
```

## Lasso L1 Regularization

The hyperparameter alpha controls the strength of the regularization. The grid search is performed to find the best value of alpha based on negative mean squared error, and then the model is fitted with the best alpha. Finally, the cross-validated R2 score is calculated and printed.

```python
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import Lasso

lasso = Lasso()

parameters = {'alpha': [1e-15, 1e-10, 1e-8, 1e-4, 1e-3,1e-2, 1, 5, 10, 20]}
lasso_regressor = GridSearchCV(lasso, 
                               parameters, 
                               scoring='neg_mean_squared_error', 
                               cv = 5,
                               )
lasso_regressor.fit(X_train, y_train)
```

```python
lasso_regressor.best_params_
```

```python
lasso_regressor.best_score_
```

```python
y_pred = lasso_regressor.predict(X_test)
```

```python
## STEP5 VERIFICA METRICHE (for Regression)
from sklearn.metrics import r2_score,mean_absolute_error,root_mean_squared_error,mean_squared_error

r2 = r2_score(y_test,y_pred)
mse = mean_squared_error(y_test,y_pred)
mae = mean_absolute_error(y_test,y_pred)
rmse = root_mean_squared_error(y_test,y_pred)

print(f'R2 score {r2}')
print(f'MSE score {mse}')
print(f'MAE score {mae}')
print(f'RMSE score {rmse}')
```

```python
r2_score = cross_val_score(lasso_regressor, X, y, scoring='r2', cv=5)
r2_score
#np.mean(r2_score)
```

## Ridge L2 Regularization

```python
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import Ridge


ridge = Ridge()

parameters = {'alpha': [1e-15, 1e-10, 1e-8, 1e-4, 1e-3,1e-2, 1, 5, 10, 20]}

ridge_regressor = GridSearchCV(ridge, 
                               parameters,
                               scoring='neg_mean_squared_error', 
                               cv=5,
                               )
ridge_regressor.fit(X_train, y_train)
```

```python
r2_score = cross_val_score(ridge, X, y, scoring='r2', cv=5)
r2_score
#np.mean(r2_score)
```

## ElasticNet (L1+L2)

 Elastic Net is a linear regression model that combines both L1 (Lasso) and L2 (Ridge) regularization. Here's an example of how you can perform hyperparameter tuning using Elastic Net with scikit-learn's ElasticNet and GridSearchCV:

```python
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import ElasticNet
from sklearn.model_selection import cross_val_score
import numpy as np

elastic_net = ElasticNet()

alpha_values = [1e-15, 1e-10, 1e-8, 1e-4, 1e-3, 1e-2, 1, 5, 10, 20]
l1_ratio_values = [0.1, 0.3, 0.5, 0.7, 0.9]

parameters = {'alpha': alpha_values, 
              'l1_ratio': l1_ratio_values,
              }

elastic_net_regressor = GridSearchCV(elastic_net, 
                                     parameters, 
                                     scoring='neg_mean_squared_error', 
                                     cv=5,
                                     )
elastic_net_regressor.fit(X_train, y_train)
```

```python
elastic_net_regressor.best_estimator_
```

```python
elastic_net_regressor.best_params_
```

```python
elastic_net_regressor.best_score_
```

```python
y_pred = elastic_net_regressor.predict(X_test)
```

```python
## STEP5 VERIFICA METRICHE (for Regression)
from sklearn.metrics import r2_score,mean_absolute_error,root_mean_squared_error,mean_squared_error

r2 = r2_score(y_test,y_pred)
mse = mean_squared_error(y_test,y_pred)
mae = mean_absolute_error(y_test,y_pred)
rmse = root_mean_squared_error(y_test,y_pred)

print(f'R2 score {r2}')
print(f'MSE score {mse}')
print(f'MAE score {mae}')
print(f'RMSE score {rmse}')
```

```python
r2_score = cross_val_score(elastic_net, X, y, scoring='r2', cv=5)
r2_score
```

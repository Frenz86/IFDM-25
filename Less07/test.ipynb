```python
## funzione per numeri pos e neg
def pos_neg(x):
    if x <0:
        print('num neg')
    elif x ==0:
        print('lo zero non è nè pos nè neg')
    else:
        print('il num è pos')
```

```python
pos_neg(0)
```

```python
var = float(input('inserisci il valore'))
pos_neg(var)
```

```python
## funzione per numeri pos e neg
def pos_neg(x):
    if x <0:
        print('num neg')
    elif x ==0:
        print('lo zero non è nè pos nè neg')
    else:
        print('il num è pos')
```

```python
pos_neg(0)
```

```python
var = float(input('inserisci il valore'))
pos_neg(var)
```

```python
def pari_dispari(k):
    if k%2==0:
        print('pari')
    else:
        print('dispari')
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-5,5,100)
y = 2*x+1

plt.plot(x, y, '-b', label='y=2x+1')
plt.title('Graph of y=2x+1')
plt.xlabel('x', color="#371C43" )
plt.ylabel('y', color="#A22F4A")
plt.legend(loc='upper left')
plt.grid()
```
